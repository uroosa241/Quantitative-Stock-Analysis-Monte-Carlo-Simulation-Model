```python
import yfinance as yf
import numpy as np

# =====================================================
# Financial Data (Apple & Tesla)
# =====================================================

apple = {
    "name": "Apple",
    "market_cap": 2.98e12,
    "revenue": 385e9,
    "net_income": 97e9,
    "operating_income": 114e9,
    "total_assets": 352e9,
    "total_liabilities": 290e9,
    "total_equity": 62e9,
    "total_debt": 110e9,
    "free_cash_flow": 95e9
}

tesla = {
    "name": "Tesla",
    "market_cap": 700e9,
    "revenue": 95e9,
    "net_income": 15e9,
    "operating_income": 12e9,
    "total_assets": 110e9,
    "total_liabilities": 45e9,
    "total_equity": 65e9,
    "total_debt": 10e9,
    "free_cash_flow": 7e9
}

# =====================================================
#  Fundamental Analysis (Ratios)
# =====================================================

def analyze(stock):
    print(f"\n📊 {stock['name']} Analysis")

    profit_margin = stock["net_income"] / stock["revenue"]
    pe_ratio = stock["market_cap"] / stock["net_income"]
    roe = stock["net_income"] / stock["total_equity"]
    debt_to_equity = stock["total_debt"] / stock["total_equity"]
    fcf_yield = stock["free_cash_flow"] / stock["market_cap"]

    print(f"Profit Margin: {profit_margin:.2%}")
    print(f"P/E Ratio: {pe_ratio:.2f}")
    print(f"ROE: {roe:.2%}")
    print(f"Debt to Equity: {debt_to_equity:.2f}")
    print(f"FCF Yield: {fcf_yield:.4%}")

# Run fundamental analysis
analyze(apple)
analyze(tesla)
```

```python
import yfinance as yf
import numpy as np

# =====================================================
#  Probability of Profit/Loss (Stock Data)
# =====================================================

data = yf.download("AAPL", start="2023-01-01", progress=False)["Close"]

returns = data.pct_change().dropna()

prob_profit = (returns > 0).mean()
prob_loss = (returns < 0).mean()

print("\n Apple Probability Analysis")
print("Profit Probability:", prob_profit)
print("Loss Probability:", prob_loss)

# =====================================================
#  Monte Carlo Simulation (GBM Model)
# =====================================================

data_mc = yf.download("AAPL", start="2020-01-01", progress=False)["Close"].dropna()

log_returns = np.log(data_mc / data_mc.shift(1)).dropna()

mu = float(log_returns.mean())
sigma = float(log_returns.std())

simulations = 1000
days = 30

final_prices = []
last_price = float(data_mc.iloc[-1])

for _ in range(simulations):
    price = last_price

    for _ in range(days):
        shock = np.random.normal(mu, sigma)
        price *= np.exp(shock)

    final_prices.append(price)

final_prices = np.array(final_prices)

print("\n Monte Carlo Results (AAPL)")
print("Expected Price:", np.mean(final_prices))
print("Best Case:", np.max(final_prices))
print("Worst Case:", np.min(final_prices))
```

```python
import numpy as np

# =====================================================
# Financial Data (Apple & Tesla)
# =====================================================

apple = {
    "name": "Apple",
    "market_cap": 2.98e12,
    "revenue": 385e9,
    "net_income": 97e9,
    "operating_income": 114e9,
    "total_assets": 352e9,
    "total_liabilities": 290e9,
    "total_equity": 62e9,
    "total_debt": 110e9,
    "free_cash_flow": 95e9
}

tesla = {
    "name": "Tesla",
    "market_cap": 700e9,
    "revenue": 95e9,
    "net_income": 15e9,
    "operating_income": 12e9,
    "total_assets": 110e9,
    "total_liabilities": 45e9,
    "total_equity": 65e9,
    "total_debt": 10e9,
    "free_cash_flow": 7e9
}



def stats(name, stock):
    values = [v for v in stock.values() if isinstance(v, (int, float))]

    mean_val = np.mean(values)
    var_val = np.var(values)
    std_val = np.std(values)

    print(f"\n==== {name} Statistics ====")
    print("Mean:", mean_val)
    print("Variance:", var_val)
    print("Std Dev:", std_val)


stats("Apple", apple)
stats("Tesla", tesla)
```

I have  built a stochastic stock price simulation model using real market data from Yahoo Finance. First, I have converted historical stock prices into daily returns to analyze how prices change over time. Then I calculated the probability of profit and loss days using an empirical probability model. For risk analysis,  I transformed returns into log returns to make the data more stable and suitable for financial modeling. Finally, I used a Monte Carlo simulation with 1000 simulations over 30 days to generate possible future stock price paths based on randomness and volatility. Overall, the model does not predict exact prices but estimates a realistic range of possible future outcomes using statistical and probabilistic methods.
