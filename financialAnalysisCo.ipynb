```python
# -----------------------------
# Financial Data
# -----------------------------

apple = {
    "name": "Apple",
    "market_cap": 2.98e12,
    "revenue": 385e9,
    "net_income": 97e9,
    "operating_income": 114e9,
    "total_assets": 352e9,
    "total_liabilities": 290e9,
    "total_equity": 62e9,
    "total_debt": 110e9,
    "free_cash_flow": 95e9
}

tesla = {
    "name": "Tesla",
    "market_cap": 700e9,
    "revenue": 95e9,
    "net_income": 15e9,
    "operating_income": 12e9,
    "total_assets": 110e9,
    "total_liabilities": 45e9,
    "total_equity": 65e9,
    "total_debt": 10e9,
    "free_cash_flow": 7e9
}


# -----------------------------
# Functions for Ratios
# -----------------------------

def analyze(stock):
    print(f"\n📊 {stock['name']} Analysis\n")

    profit_margin = stock["net_income"] / stock["revenue"]
    pe_ratio = stock["market_cap"] / stock["net_income"]
    roe = stock["net_income"] / stock["total_equity"]
    debt_to_equity = stock["total_debt"] / stock["total_equity"]
    fcf_yield = stock["free_cash_flow"] / stock["market_cap"]

    print(f"Profit Margin: {profit_margin:.2%}")
    print(f"P/E Ratio: {pe_ratio:.2f}")
    print(f"ROE: {roe:.2%}")
    print(f"Debt-to-Equity: {debt_to_equity:.2f}")
    print(f"Free Cash Flow Yield: {fcf_yield:.2%}")


# -----------------------------
# Run Analysis
# -----------------------------

analyze(apple)
analyze(tesla)
```
